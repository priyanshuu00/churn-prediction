# Explainability

Extract LR coefficients to rank top drivers.

```
import os, sys
if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')
sys.path.append(os.getcwd())
import pandas as pd
import matplotlib.pyplot as plt
from src.model.churn_classifier import ChurnModel
df = pd.read_csv('data/processed/cleaned_customer_data.csv')
model = ChurnModel()
X_train, _, y_train, _ = model.prepare_data(df)
model.train(X_train, y_train)
importance = model.get_model_coefficients()
display(importance)

# Top 10 Features Chart
importance.head(10).set_index('Feature').plot(kind='barh', color='purple', title='Top Churn Drivers (LR Coefficients)')
plt.gca().invert_yaxis()
plt.show()
```
