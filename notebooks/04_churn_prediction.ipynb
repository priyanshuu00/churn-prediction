# Churn Prediction Modeling

Train LR, evaluate Precision/Recall, output metrics.

```
import os, sys
if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')
sys.path.append(os.getcwd())
import pandas as pd
import matplotlib.pyplot as plt
from src.model.churn_classifier import ChurnModel
df = pd.read_csv('data/processed/cleaned_customer_data.csv')
model = ChurnModel()
X_train, X_test, y_train, y_test = model.prepare_data(df)
model.train(X_train, y_train)
report, auc = model.evaluate(X_test, y_test, threshold=0.4)
print(report)
print(f'AUC: {auc}')

# Visualize predicted probabilities
probs = model.model.predict_proba(X_test)[:, 1]
plt.hist(probs, bins=20, color='coral', edgecolor='black')
plt.axvline(x=0.4, color='red', linestyle='--', label='Threshold=0.4')
plt.title('Distribution of Churn Probabilities')
plt.legend()
plt.show()
```
